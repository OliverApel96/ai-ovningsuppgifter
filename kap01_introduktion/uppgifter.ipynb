#### 3. Förklara följande:
- a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och
testdata?
Träningsdata är det modellen Tränar på. Valideringsdata används för att jämföra olika modeller. Testdata rörs inte förrän allt är bestämt, och används en gång för en ärlig bedömning. Så man delar upp det för att en modell inte ska lära sig all data utantill och ge bättre resultat än vad den faktiskt gör. 

- b) Vad är k-delad korsvalidering för något?
Modellen delas in i k lika delar och tränas k gånger och varje gång är en ny del valideringsdel och resten används som träningsdata. Sedan tar man medlevärdet av allt. 

- c) Vad är root mean squared error (RMSE) för något?
Ett mått på genomsnittligt fel vid regression.
Error - för varje observation: hur mycket felade vi? (Det verkliga värdet minus predikterat)
Squared — kvadraten på varje fel. Det gör så att minustecken försvinner och stora fel straffas oproportionerligt hårt.
Mean - ta medelvärdet av de kvadrerade felen.
Root - roten ur.

- d) Vad är en hyperparameter för något? Vad är en parameter för något?
Hyperparameter är inställningar man sätter själv innan träningen, t.ex. max_depth i ett beslutsträd. Modellen kan inte lära sig den, man behöver sätta det innan. 
En parameter är något modellen räknar fram själv under träningen. I linjär regression är parametrarna koefficienterna, linjens lutning och skärningspunkt.

- f) Vad är kategorisk data och hur hanteras det? I ditt svar, använd begreppen
nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding
och ordinal encoding. 
Kategorisk data är variabler med kategorier istället för tal. Måste kodas om till siffror eftersom modeller bara räknar. Nominal har ingen rangordning (färg, land, bilmärke), Ordinal har naturlig ordning (small/medium/large, betyg), Ordinal encoding — heltal per kategori (0, 1, 2). Passar ordinal data. Fel för nominal, eftersom det bygger in en ordning som inte finns.
One-hot encoding — en 0/1-kolumn per kategori. Standardvalet för nominal data.
Dummy-variable encoding — samma som one-hot minus en kolumn, för att undvika multikollinearitet (kolumnerna summerar alltid till 1, så en är överflödig).

- h) Vad menas med principle of parsimony?
Väljer man mellan modeller som presterar likvärdigt, ta den enklaste modellen, färre variabler och parametrar.
Komplexa modeller överanpassar lättare, enkla modeller går att förklara för andra, och de kostar mindre i tid och datainsamling.

#### 4. "En modell är en förenkling av verkligheten"
En modell tar med det som är relevant för syftet och utelämnar resten — som en tunnelbanekarta, som är geografiskt felaktig men perfekt för att hitta rätt tåg. I verkligheten går rälsen huller om buller men på kartan går dom i en raklinje med stopp som är lätta att följa, så man vet vart man ska hoppa av. 

#### 5. Vad menas med att en modell är “överanpassad” eller overfitted på engelska?
En överanpassad modell har lärt sig träningsdatan utantill istället för det generella mönstret. T.ex. som en elev som memorerar gamla tentasvar men inte förstår ämnet.

#### 6/9. Högre är bättre i scikit-learn scoring, vad innebär det?
Scikit-learn har en genomgående konvention: alla scoring-mått är byggda så att ett högre värde betyder en bättre modell. Verktyg som GridSearchCV och cross_validate() kan därför alltid göra samma sak — leta upp det största värdet — oavsett vilket mått du valt.

Problemet är att felmått går åt andra hållet — lågt RMSE är bra. Lösningen är att multiplicera med −1, därav neg_mean_squared_error. MSE 100 blir −100 och MSE 400 blir −400; nu är −100 störst, och rätt modell vinner enligt samma regel som för accuracy.

#### 7. Vad är tvärsnittsdata, tidsseriedata och paneldata? Exemplifiera när respektive datakategori kan uppstå.
Tvärsnittsdata — många enheter, en tidpunkt. Ögonblicksbild. Ex: enkät bland 1 000 personer i mars, eller salary_dataset.csv med 29 olika personer. Vanligast i ML; det slumpmässig train_test_split() förutsätter.

Tidsseriedata — en enhet, många tidpunkter. Ordningen betyder något. Ex: aktiekurs dagligen i tio år, månadsomsättning, timtemperatur. Får inte splittas slumpmässigt — testsetet måste vara den senare perioden, annars predikterar man det förflutna med kunskap om framtiden.

Paneldata — många enheter över många tidpunkter. Ex: 500 företags omsättning per år 2010–2024, eller samma patienter mätta årligen. Rikast på information (man kan skilja skillnader mellan enheter från förändringar inom en enhet), men mer komplex att hantera.
